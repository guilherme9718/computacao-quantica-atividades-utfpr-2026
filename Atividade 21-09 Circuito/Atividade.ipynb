```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler
from qiskit.visualization import plot_histogram
import math

qc = QuantumCircuit(2)
qc.h(0)
qc.t(0)
qc.cx(0, 1)
qc.measure_all()

qc.draw('latex')
```

```python
sampler = StatevectorSampler()
result = sampler.run([qc], shots=1000).result()
plot_histogram(result[0].data.meas.get_counts())
```

## Análise algébrica

```python
%%latex
\begin{align}
\ket{\psi_0} = \ket{0} \otimes \ket{0} \\
\ket{\psi_1} = H\ket{0} \otimes \ket{0} = \frac{1}{\sqrt{2}}(\ket{0}+\ket{1}) \otimes \ket{0} \\
\ket{\psi_2} = \frac{1}{\sqrt{2}}(T\ket{0}+T\ket{1}) \otimes \ket{0} = \frac{1}{\sqrt{2}}\ket{0}+ (\frac{1}{2} + \frac{i}{2}) \ket{1} \otimes \ket{0} \\
\ket{\psi_3} = CX (\frac{1}{\sqrt{2}}\ket{0}+ (\frac{1}{2} + \frac{i}{2}) \ket{1} \otimes \ket{0}) \\
\ket{\psi_3} = CX (\frac{1}{\sqrt{2}}\ket{00}+ (\frac{1}{2} + \frac{i}{2}) \ket{1}) \\
\ket{\psi_3} = \frac{1}{\sqrt{2}}\ket{00}+ (\frac{1}{2} + \frac{i}{2}) \ket{11} \\
\end{align}
```

Esse estado final de dois qubits podemos verificar que a probabilidade é 50% para o bit 0 e 50% para o bit 3. Está conforme os experimentos.

```python
%%latex
\begin{align}
    p(\ket{\psi_3}) = |\frac{1}{\sqrt{2}}\ket{00}+ (\frac{1}{2} + \frac{i}{2}) \ket{11}|^2 \\
    p(\ket{\psi_3}) = \frac{1}{2}\ket{00} + \frac{1}{2}\ket{11} \\
\end{align}
```
