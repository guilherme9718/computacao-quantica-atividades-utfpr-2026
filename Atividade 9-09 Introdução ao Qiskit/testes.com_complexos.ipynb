```python
import math

class Complexo:
    def __init__(self, a, b) -> None:
        self.a = a
        self.b = b

    def modulo(self):
        return math.sqrt(self.a**2 + self.b**2)

    def theta(self):
        if self.a == 0 and self.b == 0:
            return 0
        if self.a > 0:
            return math.asin(self.b / self.modulo())
        return math.pi - math.asin(self.b / self.modulo())

    def polar(self):
        theta = self.theta()
        return self.modulo() * (math.cos(theta) + 1j * math.sin(theta))

    def polar_euler(self):
        theta = self.theta()
        return self.modulo() * math.e ** (1j * theta)

    def __str__(self) -> str:
        return f'{self.a} + {self.b}i'

    def obter_info(self):
        print(self)
        print(f'modulo = {self.modulo()}')
        print(f'theta(rad) = {self.theta()}')
        print(f'theta(graus) = {self.theta() * 180.0 / math.pi}')
        print(f'polar = {self.polar()}')
        print(f'polar_euler = {self.polar_euler()}')
        print()
```

```python
Complexo(3, 4).obter_info() # Quadrante 1
Complexo(-3, 4).obter_info() # Quadrante 2
Complexo(-3, -4).obter_info() # Quadrante 3
Complexo(3, -4).obter_info() # Quadrante 4
```
