```python
from numpy import sqrt
import numpy as np
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_histogram
import math
import cmath

sv=Statevector([1 / sqrt(2), 1j / sqrt(2)])

display(sv.is_valid())
```

```python
display(sv.draw("latex"))
```

```python
stats = sv.sample_counts(1000)
plot_histogram(stats)
```

```python
def remove_parte_imaginaria(a, b, c, d):

  norm_0 = sqrt(a**2 + b**2)
  norm_1 = sqrt(c**2 + d**2)

  #theta para o 0
  theta = math.asin(b/norm_0)
  #zeta para o 1
  zeta = math.asin(d/norm_1)
  #|0> = a + bi
  negative_flag_0 = 0
  if a < 0:
    negative_flag_0 = math.pi

  negative_flag_1 = 0
  if c < 0:
    negative_flag_1 = math.pi

  state = Statevector([a + b*1j, c + d*1j])
  #|1> = c + di

  psi = Statevector([norm_0, norm_1 * (math.e**(1j*(zeta + negative_flag_1 - (theta + negative_flag_0))))])
  
  print("saida eh valida?")
  display(psi.is_valid())
  print("entrada:\n")
  display(state.draw("latex"))
  print("saida:\n")
  display(psi.draw("latex"))
  print("sao equivalentes?:")
  display(psi.equiv(state))
  print("bloch  entrada:\n")
  display(state.draw("bloch"))
  print("bloch  saida:\n")
  display(psi.draw("bloch"))
  print("amostragem 1000:\n")
  stats = psi.sample_counts(1000)
  display(plot_histogram(stats))



  return psi
```
